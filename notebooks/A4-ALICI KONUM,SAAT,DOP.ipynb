```python
# başlangıç ile bitiş satırı arası verileri toplar, satırın tamamını girdi olarak yaz, boşluklar ile beraber kopyala
def get_data(file_path, start_line, end_line):
    data = []
    collecting = False

    with open(file_path, "r") as f:
        for line in f:
            if line.rstrip() == start_line:
                collecting = True

            if line.rstrip() == end_line:
                break

            if collecting:
                data.append(line.rstrip())

    return data
```

### **lsq fonksiyon**

```python
# P ve sat_coords önceden geliyor
def spp_lsq(P, sat_coords, rec_init, max_iter=40, tol=1e-4): # max iterasyon ve tolerans default = 10, 0.001metre = 1milimetre
    n = len(P) # pseudo range sayısı 9du 8e düşürdük
    P = np.asarray(P, dtype=float) 
    sat_coords = np.asarray(sat_coords, dtype=float) # uydu koord.
    X_rec = np.asarray(rec_init, dtype=float).copy() # alıcı aprox koord.
    c_dt_rec = 0.0 # Alıcı saat hatasının mesafesel karşılığı [m] = 0 kabul ile başlıyoruz iterasyona

    for iteration in range(max_iter): # max iter 40
        A = np.zeros((n, 4)) 
        l = np.zeros(n)

        for i in range(n): # her pseudorange için
            d = sat_coords[i] - X_rec # geometrik mesafe xyz, X_rec güncel geliyor
            rho = np.linalg.norm(d) # geometrik mesafe vektör = uzunluk
            A[i, :3] = -d / rho # Tasarım matrisi
            A[i, 3] = 1.0
            l[i] = P[i] - rho - c_dt_rec # Ölçü farkı, c_dt_rec güncel geliyor
        print("ITERASYON ve A ve l : ",iteration,A,l)
        # LSQ çözümü
        dx, residuals, _, _ = np.linalg.lstsq(A, l) #dx, düzeltmeler
        print("dx ve residuals: ",dx, residuals)
        X_rec += dx[:3] # dx 0,1,2 elementlerini ekle X_rec için başlangıçta rec_initle aynı
        c_dt_rec += dx[3] # c_dt 3.element
        if np.linalg.norm(dx[:3]) < tol: #koord vektör uzunluk 1mmden kısa ise tamam
            break

    # Son çözümde tasarım matrisini ve kalıntıları yeniden hesapla
    A = np.zeros((n, 4))
    v = np.zeros(n)
    for i in range(n): # range(8) her bir uydu için son döngü, vleri bul
        d = sat_coords[i] - X_rec
        rho = np.linalg.norm(d)
        A[i, :3] = -d / rho
        A[i, 3] = 1.0
        v[i] = P[i] - rho - c_dt_rec
        
    Q = np.linalg.inv(A.T @ A) # DOP hesabı için kofaktör matrisi
    dt_rec = c_dt_rec / c # Alıcı saat hatası [s]
    return X_rec, dt_rec, Q, v
```

### **dop çıktı**

```python
def calculate_dops(Q):
    GDOP = np.sqrt(np.trace(Q))
    PDOP = np.sqrt(np.trace(Q[:3, :3]))
    TDOP = np.sqrt(Q[3, 3])
    return GDOP, PDOP, TDOP

GDOP_P1, PDOP_P1, TDOP_P1 = calculate_dops(Q_P1)
GDOP_P2, PDOP_P2, TDOP_P2 = calculate_dops(Q_P2)

print("       P1       P2")
print(f"GDOP  {GDOP_P1:.4f}   {GDOP_P2:.4f}")
print(f"PDOP  {PDOP_P1:.4f}   {PDOP_P2:.4f}")
print(f"TDOP  {TDOP_P1:.4f}   {TDOP_P2:.4f}")
```
