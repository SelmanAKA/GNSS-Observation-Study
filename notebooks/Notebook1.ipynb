```python
# başlangıç ile bitiş satırı arası verileri toplar, satırın tamamını girdi olarak yaz, boşluklar ile beraber kopyala
def get_epoch_data(file_path, start_epoch, end_epoch):
    data = []
    collecting = False

    with open(file_path, "r") as f:
        for line in f:
            if line.rstrip() == start_epoch:
                collecting = True

            if line.rstrip() == end_epoch:
                break

            if collecting:
                data.append(line.rstrip())

    return data
```

## A2.  Calculate the multipath error on P1, P2 code pseudoranges for all satellites in given time.

### Giriş

- GNSS sinyalleri uydudan alıcıya iki şekilde ulaşır doğrudan ve yansıyarak. Yansıtıcı yüzeylerden (bina, su vb.) gelen sinyaller alıcının hatalı ölçüm yapmasına sebep olur. Bu sinyal yansıma hatası (multipath error) olarak adlandırılır.
- Sözde menzil (Pseudorange) uydudan alıcıya gelen sinyalleri arada geçen zamanla çarparak bulunur. Sinyaller c=ışık hızında hareket eder. P=c.t
- GNSS Uydularından farklı dalga boylarında farklı bilgiler alınır. L1 ve L2 frekanslarında faz bilgileri ile P1 ve P2 kod bilgileri elde edilir. Kod bilgileri metre ile ifade edilir ve düşük hassasiyetli gözlemler yapılır. Faz bilgileri çevirim olarak saklanır ve metreye çevrilmelidir.

### GF Mp1 Mp2

- 

```python
# ışık hızı m/s
c = 299792458  
# frekanslar Hz 
f1 = 1575.42 * 1000000
f2 = 1227.60 * 1000000
# lambda c/f metre/saniye / çevirim/saniye ==> metre/çevirim lambdanın birimi
lambda1 = c/f1
lambda2 = c/f2
```

```python
lambda1, lambda2
```

```python
for i in range(9):
    sat_name = globals()[f"sat{i+1}"][0] #G0    
    sat_L1 = globals()[f"sat{i+1}"][3] #L1
    sat_L2 = globals()[f"sat{i+1}"][6] #L2
    GF = lambda1 * sat_L1 - lambda2 * sat_L2 # GF Geometry free faz kombinasyonu
    print(sat_name, GF)
```

```python
# MP1
for i in range(9):
    sat_name = globals()[f"sat{i+1}"][0] #G01
    sat_P1 = globals()[f"sat{i+1}"][2] #P1
    sat_L1 = globals()[f"sat{i+1}"][3] #L1
    sat_P2 = globals()[f"sat{i+1}"][5] #P2
    sat_L2 = globals()[f"sat{i+1}"][6] #L2
    Phi1 = sat_L1 * lambda1 # L sinyali çevirim lambda metre /çevirim fi1 metre biriminde elde edilir.
    Phi2 = sat_L2 * lambda2
    MP1 = sat_P1 - ((f1*f1+f2*f2)/(f1*f1-f2*f2))*Phi1 + ((2*f2*f2)/(f1*f1-f2*f2))*Phi2
    print(sat_name, MP1)
```

```python
#MP2
for i in range(9):
    sat_name = globals()[f"sat{i+1}"][0] #G01
    sat_P1 = globals()[f"sat{i+1}"][2] #P1
    sat_L1 = globals()[f"sat{i+1}"][3] #L1
    sat_P2 = globals()[f"sat{i+1}"][5] #P2
    sat_L2 = globals()[f"sat{i+1}"][6] #L2
    Phi1 = sat_L1 * lambda1 # L sinyali çevirim lambda metre /çevirim fi1 metre biriminde elde edilir.
    Phi2 = sat_L2 * lambda2
    MP2 = sat_P2 - (2*f1*f1 / (f1*f1 - f2*f2)) * Phi1 + ((f1*f1 + f2*f2) / (f1*f1 - f2*f2)) * Phi2
    print(sat_name, MP2)
```
