```python
# başlangıç ile bitiş epoğu arası verileri toplar. satırın tamamını girdi olarak yaz.
def get_epoch_data(file_path, start_epoch, end_epoch):
    data = []
    collecting = False

    with open(file_path, "r") as f:
        for line in f:
            if line.rstrip() == start_epoch:
                collecting = True

            if line.rstrip() == end_epoch:
                break

            if collecting:
                data.append(line.rstrip())

    return data
```
